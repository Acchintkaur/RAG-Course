```python
from langchain_classic.storage import LocalFileStore
from langchain_classic.embeddings.cache import CacheBackedEmbeddings
from langchain_openai import OpenAIEmbeddings

embeddings_model = OpenAIEmbeddings(model="text-embedding-3-small")

doc_store = LocalFileStore(root_path="./embedding_cache/documents")
query_store = LocalFileStore(root_path="./embedding_cache/queries")

cached_embeddings = CacheBackedEmbeddings.from_bytes_store(
    underlying_embeddings=embeddings_model,
    document_embedding_cache=doc_store,   # required
    query_embedding_cache=query_store,    # optional, enables query caching
    namespace="acchintTry"
)

query = "What is the use of context-aware cognitive augmentation?"
query_vector = cached_embeddings.embed_query(query)
query_vector
```

```python
# Step 3: Query Embedding
query_vector_twice = cached_embeddings.embed_query(query)
query_vector_twice
```

```python
import numpy as np
np.allclose(query_vector, query_vector_twice)
```
