### WebBaseLoader

```python
# WebBaseLoader downloads a web page and converts its HTML into a LangChain Document.
from langchain_community.document_loaders import WebBaseLoader

# Keep the URLs in a list so we can load several web pages together.
urls = [
    "https://docs.python.org/3/tutorial/",
    "https://docs.python.org/3/library/pathlib.html",
    "https://docs.python.org/3/library/json.html",
]

# Create one loader for all the URLs.
loader = WebBaseLoader(web_paths=urls)

# load() visits every URL and returns one Document for each page.
documents = loader.load()

# Check how many pages were loaded.
print(f"Loaded {len(documents)} web pages")

for doc in documents:
    print("Source:", doc.metadata.get("source"))
    print(doc)
```
