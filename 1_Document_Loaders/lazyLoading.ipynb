### Lazy Loading (lazy_load())
What It Is?
lazy_load() fetches documents one by one on demand. 
Instead of returning a list in memory, it returns a Python generator.  

Why It Is Needed? 
1. Prevents Out-Of-Memory (OOM) Errors: If you try to load a 10 GB dataset into 4 GB of RAM using load(), your program will crash.
2. Scalability: It enables you to load, process (chunk), embed, and store documents in batches, releasing memory continuously as it iterates.

```python
from pathlib import Path
from langchain_community.document_loaders import DirectoryLoader, TextLoader

data_folder = Path("data/text_files")

loader = DirectoryLoader(
    str(data_folder),
    glob="*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
)

# lazy_load() returns a generator object (doesn't read everything immediately)
doc_generator = loader.lazy_load()

doc_generator
```

```python
for document in doc_generator:
    print(document.metadata["source"])
```
